## 11. Reto 10 — Contar elementos

Crea `contar_elementos(valores)` utilizando un contador y un recorrido. Retorna la cantidad de elementos.

Prueba con `[10, 20, 30, 40, 50]`. Resultado esperado: `5`.

Pregunta: ¿qué diferencia existe entre esta función y `len()`?

````python
def contar_elementos(valores):
    # TODO: inicializa un contador.
    contador = 0
    # TODO: aumenta el contador por cada elemento.
    for _ in valores:
        contador += 1
    # TODO: retorna el contador.
    return contador
````

## 13. Reto 12 — Generar un resumen por clasificación

Implementa `generar_resumen(valores)` para contar cuántos elementos son `ALTO`, `NORMAL` y `BAJO`. Debe reutilizar `clasificar_valor()` y retornar los resultados.

Requisito: inicializa los contadores antes del ciclo.

````python
def generar_resumen(valores):
    altos = 0
    normales = 0
    bajos = 0

    # TODO: recorre valores y clasifica cada uno.
    for valor in valores:
        categoria = clasificar_valor(valor)
        # TODO: actualiza el contador correspondiente.
        if categoria == "ALTO":
            altos += 1
        elif categoria == "NORMAL":
            normales += 1
        elif categoria == "BAJO":
            bajos += 1

    # TODO: retorna altos, normales, bajos.
    return altos, normales, bajos
````

¿Qué problema resuelve una lista? Agrupa y organiza muchos datos en una sola variable en lugar de usar una para cada valor.

¿Qué diferencia hay entre índice y valor? El índice es la posición numérica del elemento (comienza en 0) y el valor es el contenido que hay en esa posición.

¿Qué diferencia hay entre len() y el último índice? len() da el total de elementos (desde 1), mientras que el último índice es ese total menos uno (len - 1).

¿Cuándo conviene recorrer por índice? Cuando necesitas modificar o reemplazar los elementos originales dentro de la lista.

¿Por qué pasar una lista como parámetro? Para que la función sea reutilizable y pueda procesar cualquier conjunto de datos.

¿Por qué reutilizar clasificar_valor()? Evita duplicar código y asegura que todos usen la misma regla de negocio si esta cambia.

¿Qué relación existe entre listas, contadores y acumuladores? La lista da los datos, el contador mide cuántas veces se cumple una condición y el acumulador suma los valores numéricos.

¿Qué nueva capacidad obtiene DataLab v0.8 frente a DataLab v0.7? Permite procesar colecciones dinámicas de datos (listas) para generar resúmenes, sumas, promedios y controles automáticos.
